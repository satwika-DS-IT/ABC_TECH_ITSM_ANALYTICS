# ABC TECH ITSM [PRCL-0012] - CLIENT PROJECT

## Project Statement

1) ABC Tech has been experiencing challenges in its IT Service Management (ITSM) incident management process, with customer feedback indicating that incident management performance needs improvement.

2) The organization maintains historical incident records containing information about incident priorities, impact, urgency, categories, configuration items, timestamps, and other incident-related attributes.

3) The objective of this project is to analyze the historical ITSM incident data and develop a machine learning classification model to identify high-priority incidents (Priority 1 and Priority 2). The analysis will also explore incident trends and other patterns that may help improve incident handling, resource allocation, and operational decision-making.

4) The project will follow a structured data science workflow, including data understanding, exploratory data analysis, data preprocessing, feature engineering, model development, and model evaluation.


## Project Objectives

The key objectives of this project are:

1. **High-Priority Incident Prediction:** Develop a classification model to identify incidents that fall under Priority 1 and Priority 2 so that appropriate preventive or corrective measures can be taken.

2. **Incident Volume Forecasting:** Analyze and forecast incident volumes across different fields on a quarterly and annual basis to identify trends and support planning.

3. **Automatic Priority and Department Tagging:** Explore how machine learning can assist in assigning appropriate priorities and departments to incidents, helping reduce unnecessary reassignment and delays.

4. **RFC and ITSM Asset Analysis:** Investigate the prediction of Requests for Change (RFCs) and potential ITSM asset failures or misconfigurations.

5. **Data-Driven Recommendations:** Derive useful insights from historical incident data and recommend ways to improve incident management and operational efficiency.

## DATA PREPROCESSING

```python
# Data Preprocessing
df_model = df[df["Priority"].notna()].copy()
df_model["High_Priority"] = (
    df_model["Priority"].isin([1, 2])
).astype(int)
print("Dataset shape after removing missing Priority:")
print(df_model.shape)
print("\nTarget Distribution:")
display(df_model["High_Priority"].value_counts().rename({
    0: "Other Priority (3, 4, 5)",
    1: "High Priority (1, 2)"
}))
print("\nTarget Percentage:")
display(
    (df_model["High_Priority"].value_counts(normalize=True) * 100).round(2)
)
```

```python
# Identify columns for exclusion
columns_to_drop = [
    "Incident_ID",
    "Priority",
    "number_cnt",
    "Alert_Status",
    "Status",
    "Reopen_Time",
    "Resolved_Time",
    "Close_Time",
    "Handle_Time_hrs",
    "Closure_Code",
    "No_of_Reassignments",
    "No_of_Related_Interactions",
    "Related_Interaction",
    "No_of_Related_Incidents",
    "No_of_Related_Changes",
    "Related_Change",
    "KB_number"
]
print("Columns selected for exclusion:")
print(columns_to_drop)
print("\nNumber of columns selected:", len(columns_to_drop))
```

```python
# Create initial modelling dataset
df_model = df_model.drop(columns=columns_to_drop)
print("Modelling dataset shape:", df_model.shape)
print("\nRemaining columns:")
print(df_model.columns.tolist())
print("\nMissing values in remaining columns:")
display(df_model.isnull().sum())
```

```python
# Handle missing categorical values
df_model["CI_Cat"] = df_model["CI_Cat"].fillna("Unknown")
df_model["CI_Subcat"] = df_model["CI_Subcat"].fillna("Unknown")
print("Missing values after treatment:")
display(df_model.isnull().sum())
print("\nUpdated dataset shape:", df_model.shape)
```

```python
# Extract date and time features
df_model["Open_Time"] = pd.to_datetime(
    df_model["Open_Time"],
    format="%d-%m-%Y %H:%M",
    errors="coerce"
)
df_model["Open_Year"] = df_model["Open_Time"].dt.year
df_model["Open_Month"] = df_model["Open_Time"].dt.month
df_model["Open_Day"] = df_model["Open_Time"].dt.day
df_model["Open_Weekday"] = df_model["Open_Time"].dt.dayofweek
df_model["Open_Hour"] = df_model["Open_Time"].dt.hour
# Remove original timestamp
df_model.drop(columns=["Open_Time"], inplace=True)
print("Updated dataset shape:", df_model.shape)
print("\nRemaining missing values:")
display(df_model.isnull().sum())
display(df_model.head())
```

```python
# Encode categorical features
# Frequency encoding for high-cardinality columns
for col in ["CI_Name", "WBS"]:
    freq = df_model[col].value_counts(normalize=True)
    df_model[col] = df_model[col].map(freq)
# One-hot encoding for remaining categorical columns
categorical_cols = [
    "CI_Cat",
    "CI_Subcat",
    "Impact",
    "Urgency",
    "Category"
]
df_model = pd.get_dummies(
    df_model,
    columns=categorical_cols,
    drop_first=True,
    dtype=int
)
print("Dataset shape after encoding:", df_model.shape)
print("\nData types:")
display(df_model.dtypes.value_counts())
display(df_model.head())
```

```python
# Feature selection and train-test split
from sklearn.model_selection import train_test_split
# Separate features and target
X = df_model.drop(
    columns=["High_Priority", "Impact_2", "Impact_3",
             "Impact_4", "Impact_5",
             "Urgency_2", "Urgency_3", "Urgency_4",
             "Urgency_5", "Urgency_5 - Very Low"],
    errors="ignore"
)
y = df_model["High_Priority"]
# Stratified 80:20 split
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
print("Training features:", X_train.shape)
print("Testing features:", X_test.shape)
print("\nTraining target distribution:")
print(y_train.value_counts())
print("\nTesting target distribution:")
print(y_test.value_counts())
```

```python
# Rebuild raw features from the original dataset
model_data = df[df["Priority"].notna()].copy()
model_data["High_Priority"] = (
    model_data["Priority"].isin([1, 2])
).astype(int)
# Select features available at incident registration
X_raw = model_data[
    ["CI_Name", "CI_Cat", "CI_Subcat", "WBS", "Category", "Open_Time"]
].copy()
y = model_data["High_Priority"]
# Handle missing categorical values
X_raw = X_raw.fillna({
    "CI_Cat": "Unknown",
    "CI_Subcat": "Unknown"
})
# Extract date features
X_raw["Open_Time"] = pd.to_datetime(
    X_raw["Open_Time"],
    format="%d-%m-%Y %H:%M",
    errors="coerce"
)
X_raw["Open_Year"] = X_raw["Open_Time"].dt.year
X_raw["Open_Month"] = X_raw["Open_Time"].dt.month
X_raw["Open_Day"] = X_raw["Open_Time"].dt.day
X_raw["Open_Weekday"] = X_raw["Open_Time"].dt.dayofweek
X_raw["Open_Hour"] = X_raw["Open_Time"].dt.hour
X_raw.drop(columns=["Open_Time"], inplace=True)
# Split before encoding
X_train_raw, X_test_raw, y_train, y_test = train_test_split(
    X_raw,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
# Frequency encoding learned only from training data
for col in ["CI_Name", "WBS"]:
    freq = X_train_raw[col].value_counts(normalize=True)
    X_train_raw[col] = X_train_raw[col].map(freq)
    X_test_raw[col] = X_test_raw[col].map(freq).fillna(0)
# One-hot encode remaining categorical features
categorical_cols = ["CI_Cat", "CI_Subcat", "Category"]
X_train = pd.get_dummies(
    X_train_raw, columns=categorical_cols, dtype=int
)
X_test = pd.get_dummies(
    X_test_raw, columns=categorical_cols, dtype=int
)
# Align test columns with training columns
X_test = X_test.reindex(columns=X_train.columns, fill_value=0)
print("Training shape:", X_train.shape)
print("Testing shape:", X_test.shape)
print("\nTraining positive cases:", y_train.sum())
print("Testing positive cases:", y_test.sum())
```

## RANDOM FOREST MODEL TRAINING

```python
# Model Creation
from sklearn.ensemble import RandomForestClassifier
# Initialize model
rf_model = RandomForestClassifier(
    n_estimators=200,
    class_weight="balanced",
    random_state=42,
    n_jobs=-1
)
# Train model
rf_model.fit(X_train, y_train)
print("Random Forest model trained successfully!")
```

```python
# Model Evaluation

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)

# Generate predictions
y_pred = rf_model.predict(X_test)

# Calculate evaluation metrics
print("Model Evaluation Results")
print("------------------------")
print("Accuracy:", round(accuracy_score(y_test, y_pred), 4))
print("Precision:", round(precision_score(y_test, y_pred, zero_division=0), 4))
print("Recall:", round(recall_score(y_test, y_pred, zero_division=0), 4))
print("F1 Score:", round(f1_score(y_test, y_pred, zero_division=0), 4))
print("\nClassification Report:")
print(classification_report(y_test, y_pred, zero_division=0))
print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))
```

```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import precision_score, recall_score, f1_score
from sklearn.ensemble import RandomForestClassifier
import pandas as pd

# Create validation data from training data
X_subtrain, X_val, y_subtrain, y_val = train_test_split(
    X_train,
    y_train,
    test_size=0.20,
    random_state=42,
    stratify=y_train
)

# Train a validation model
validation_model = RandomForestClassifier(
    n_estimators=200,
    class_weight="balanced",
    random_state=42,
    n_jobs=-1
)

validation_model.fit(X_subtrain, y_subtrain)

# Get high-priority probabilities
val_probabilities = validation_model.predict_proba(X_val)[:, 1]

# Compare different thresholds
results = []

for threshold in [0.50, 0.40, 0.30, 0.20, 0.10]:
    predictions = (val_probabilities >= threshold).astype(int)

    results.append({
        "Threshold": threshold,
        "Precision": round(precision_score(y_val, predictions, zero_division=0), 3),
        "Recall": round(recall_score(y_val, predictions, zero_division=0), 3),
        "F1_Score": round(f1_score(y_val, predictions, zero_division=0), 3)
    })

threshold_results = pd.DataFrame(results)
display(threshold_results)
```

```python
# Objective 3: Data preparation and feature engineering

# Create a separate modelling dataset
tagging_df = priority_df.copy()

# 1. Clean categorical columns
categorical_columns = [
    "CI_Name",
    "CI_Cat",
    "CI_Subcat",
    "WBS",
    "Category"
]

for col in categorical_columns:
    tagging_df[col] = tagging_df[col].fillna("Unknown").astype(str).str.strip()


# 2. Clean Impact and Urgency
tagging_df["Impact"] = pd.to_numeric(
    tagging_df["Impact"],
    errors="coerce"
)

tagging_df["Urgency"] = (
    tagging_df["Urgency"]
    .astype(str)
    .str.strip()
    .replace({"5 - Very Low": "5"})
)

tagging_df["Urgency"] = pd.to_numeric(
    tagging_df["Urgency"],
    errors="coerce"
)


# 3. Extract date and time features
tagging_df["Open_Year"] = tagging_df["Open_Time"].dt.year
tagging_df["Open_Month"] = tagging_df["Open_Time"].dt.month
tagging_df["Open_Day"] = tagging_df["Open_Time"].dt.day
tagging_df["Open_Weekday"] = tagging_df["Open_Time"].dt.dayofweek
tagging_df["Open_Hour"] = tagging_df["Open_Time"].dt.hour


# 4. Convert target variable to integer
tagging_df["Priority"] = tagging_df["Priority"].astype(int)


# 5. Display preparation results
print("Prepared dataset shape:", tagging_df.shape)

print("\nPriority distribution:")
display(tagging_df["Priority"].value_counts().sort_index())

print("\nMissing values in selected features:")
display(
    tagging_df[
        categorical_columns + [
            "Impact",
            "Urgency",
            "Open_Year",
            "Open_Month",
            "Open_Day",
            "Open_Weekday",
            "Open_Hour"
        ]
    ].isnull().sum()
)

print("\nCleaned Urgency distribution:")
display(tagging_df["Urgency"].value_counts(dropna=False).sort_index())

print("\nSample prepared records:")
display(tagging_df.head())
```

```python
# Objective 3: Prepare training and testing datasets

from sklearn.model_selection import train_test_split

# 1. Keep Priority 1 records separate because there are only 3
priority1_records = tagging_df[
    tagging_df["Priority"] == 1
].copy()

# 2. Prepare modelling dataset with Priority 2 to 5
modeling_df = tagging_df[
    tagging_df["Priority"].isin([2, 3, 4, 5])
].copy()

print("Priority 1 records kept separately:", len(priority1_records))
print("Records available for modelling:", len(modeling_df))


# 3. Define common features
base_features = [
    "CI_Name",
    "CI_Cat",
    "CI_Subcat",
    "WBS",
    "Category",
    "Open_Year",
    "Open_Month",
    "Open_Day",
    "Open_Weekday",
    "Open_Hour"
]

# Model A: Include Impact and Urgency
features_A = base_features + ["Impact", "Urgency"]

# Model B: Exclude Impact and Urgency
features_B = base_features

# Target variable
y = modeling_df["Priority"]


# 4. Split dataset using stratification
train_indices, test_indices = train_test_split(
    modeling_df.index,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# Prepare Model A datasets
X_A_train = modeling_df.loc[train_indices, features_A]
X_A_test = modeling_df.loc[test_indices, features_A]

# Prepare Model B datasets
X_B_train = modeling_df.loc[train_indices, features_B]
X_B_test = modeling_df.loc[test_indices, features_B]

# Prepare target datasets
y_train = modeling_df.loc[train_indices, "Priority"]
y_test = modeling_df.loc[test_indices, "Priority"]


# 5. Display split information
print("\nTraining dataset shape:", X_A_train.shape)
print("Testing dataset shape:", X_A_test.shape)

print("\nTraining target distribution:")
display(y_train.value_counts().sort_index())

print("\nTesting target distribution:")
display(y_test.value_counts().sort_index())

print("\nModel A feature count:", len(features_A))
print("Model B feature count:", len(features_B))
```

```python
# Objective 3: Feature encoding and preprocessing

from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestClassifier

# 1. Create copies of training and testing features
X_A_train_processed = X_A_train.copy()
X_A_test_processed = X_A_test.copy()

X_B_train_processed = X_B_train.copy()
X_B_test_processed = X_B_test.copy()


# 2. Frequency encoding for CI_Name and WBS
frequency_columns = ["CI_Name", "WBS"]

frequency_mappings = {}

for col in frequency_columns:

    frequency_map = (
        X_A_train[col]
        .value_counts(normalize=True)
        .to_dict()
    )

    frequency_mappings[col] = frequency_map

    # Apply training frequency mapping
    X_A_train_processed[col] = (
        X_A_train[col].map(frequency_map).fillna(0)
    )

    X_B_train_processed[col] = (
        X_B_train[col].map(frequency_map).fillna(0)
    )

    # Apply the same training mapping to test data
    X_A_test_processed[col] = (
        X_A_test[col].map(frequency_map).fillna(0)
    )

    X_B_test_processed[col] = (
        X_B_test[col].map(frequency_map).fillna(0)
    )


# 3. Define categorical and numerical features
categorical_features = [
    "CI_Cat",
    "CI_Subcat",
    "Category"
]

numerical_features_A = [
    "Open_Year",
    "Open_Month",
    "Open_Day",
    "Open_Weekday",
    "Open_Hour",
    "Impact",
    "Urgency"
]

numerical_features_B = [
    "Open_Year",
    "Open_Month",
    "Open_Day",
    "Open_Weekday",
    "Open_Hour"
]

# Frequency-encoded columns are numerical too
numerical_features_A += frequency_columns
numerical_features_B += frequency_columns


# 4. Create preprocessing pipelines
preprocessor_A = ColumnTransformer(
    transformers=[
        (
            "categorical",
            OneHotEncoder(handle_unknown="ignore"),
            categorical_features
        ),
        (
            "numerical",
            "passthrough",
            numerical_features_A
        )
    ]
)

preprocessor_B = ColumnTransformer(
    transformers=[
        (
            "categorical",
            OneHotEncoder(handle_unknown="ignore"),
            categorical_features
        ),
        (
            "numerical",
            "passthrough",
            numerical_features_B
        )
    ]
)


# 5. Check frequency encoding results
print("Frequency encoding completed.")

for col in frequency_columns:
    print(f"\n{col} - unique training categories:",
          len(frequency_mappings[col]))

print("\nModel A training shape:", X_A_train_processed.shape)
print("Model A testing shape:", X_A_test_processed.shape)

print("\nModel B training shape:", X_B_train_processed.shape)
print("Model B testing shape:", X_B_test_processed.shape)

print("\nPreprocessing pipelines created successfully.")
```

```python
# Objective 3: Train Random Forest classification models

from sklearn.ensemble import RandomForestClassifier
from sklearn.pipeline import Pipeline

# 1. Define Model A
rf_model_A = Pipeline([
    ("preprocessing", preprocessor_A),
    ("classifier", RandomForestClassifier(
        n_estimators=200,
        class_weight="balanced",
        random_state=42,
        n_jobs=-1
    ))
])

# 2. Define Model B
rf_model_B = Pipeline([
    ("preprocessing", preprocessor_B),
    ("classifier", RandomForestClassifier(
        n_estimators=200,
        class_weight="balanced",
        random_state=42,
        n_jobs=-1
    ))
])


# 3. Train Model A
print("Training Model A: With Impact and Urgency...")

rf_model_A.fit(X_A_train_processed, y_train)

print("Model A training completed.")


# 4. Train Model B
print("\nTraining Model B: Without Impact and Urgency...")

rf_model_B.fit(X_B_train_processed, y_train)

print("Model B training completed.")


# 5. Generate predictions
y_pred_A = rf_model_A.predict(X_A_test_processed)
y_pred_B = rf_model_B.predict(X_B_test_processed)

print("\nPredictions generated successfully.")

print("Model A predictions:", len(y_pred_A))
print("Model B predictions:", len(y_pred_B))
```

```python
# Objective 3: Evaluate Random Forest models

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)

# Define evaluation function
def evaluate_model(model_name, y_true, y_pred):

    print("\n" + "=" * 65)
    print(model_name)
    print("=" * 65)

    print("Accuracy:",
          round(accuracy_score(y_true, y_pred), 4))

    print("Precision (weighted):",
          round(precision_score(
              y_true, y_pred,
              average="weighted",
              zero_division=0
          ), 4))

    print("Recall (weighted):",
          round(recall_score(
              y_true, y_pred,
              average="weighted",
              zero_division=0
          ), 4))

    print("F1-score (weighted):",
          round(f1_score(
              y_true, y_pred,
              average="weighted",
              zero_division=0
          ), 4))

    print("\nClassification Report:")
    print(
        classification_report(
            y_true,
            y_pred,
            labels=[2, 3, 4, 5],
            zero_division=0
        )
    )

    print("Confusion Matrix:")
    display(
        pd.DataFrame(
            confusion_matrix(
                y_true,
                y_pred,
                labels=[2, 3, 4, 5]
            ),
            index=["Actual 2", "Actual 3", "Actual 4", "Actual 5"],
            columns=["Predicted 2", "Predicted 3",
                     "Predicted 4", "Predicted 5"]
        )
    )


# Evaluate Model A
evaluate_model(
    "MODEL A - WITH IMPACT AND URGENCY",
    y_test,
    y_pred_A
)


# Evaluate Model B
evaluate_model(
    "MODEL B - WITHOUT IMPACT AND URGENCY",
    y_test,
    y_pred_B
)
```
